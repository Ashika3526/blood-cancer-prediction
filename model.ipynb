```python
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
df=pd.read_csv(r"D:\AK infopark\blood_cancer_with_binary_labels.csv")
print(df.tail())
df['label'] = df['label'].map({'Blood_Cancer': 1, 'Normal': 0})
print(df['label'].value_counts())
x = df[['RBC count', 'Haemoglobin', 'Platelet count', 'ECG / Heart rate', 'LDH']]  
y = df[['label']]
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy}")
def predict_from_user():
    print("\nEnter the following medical values for prediction:")
    rbc = float(input("RBC count: "))
    hemoglobin = float(input("Haemoglobin: "))
    platelet = float(input("Platelet count: "))
    ecg_heart_rate = float(input("ECG / Heart rate: "))
    ldh = float(input("LDH: "))
    user_input = pd.DataFrame([[rbc, hemoglobin, platelet, ecg_heart_rate, ldh]],columns=['RBC count', 'Haemoglobin', 'Platelet count', 'ECG / Heart rate', 'LDH'])
    prediction = model.predict(user_input)
    if prediction[0] == 1:
        print("🔴 The patient is likely to have Blood Cancer.")
    else:
        print("🟢 The patient is likely to be Normal.")
predict_from_user()


```
